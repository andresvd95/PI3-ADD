# Reconocimiento del dataset Titanic para la necesidad analítica

Este notebook obtiene el dataset `BIT/titanic-dataset` desde Hugging Face y revisa su estructura para formular un proyecto descriptivo sobre la supervivencia observada entre grupos de pasajeros.

```python
from pathlib import Path

import pandas as pd
from datasets import load_dataset
from IPython.display import display
```

## Obtención de los datos

Se utiliza el conjunto completo disponible en el split `train`. No se toma una muestra.

```python
DATASET_ID = "BIT/titanic-dataset"
SPLIT = "train"
DATASET_URL = "https://huggingface.co/datasets/BIT/titanic-dataset"
FECHA_CONSULTA = "23 de octubre de 2026"

dataset = load_dataset(DATASET_ID, split=SPLIT)
titanic_df = dataset.to_pandas().copy()
print(f"Dataset cargado: {DATASET_ID} ({SPLIT})")
```

## Registros y dimensiones

```python
display(titanic_df.head())
print(f"Filas: {titanic_df.shape[0]}")
print(f"Columnas: {titanic_df.shape[1]}")
```

Cada fila representa a una persona registrada como pasajera del Titanic. Para la necesidad propuesta se usarán principalmente `survived`, `pclass`, `sex`, `age`, `fare` y `embark_town`, porque permiten describir y comparar la supervivencia observada según características demográficas y del viaje. El análisis será descriptivo y no establecerá causas.

## Columnas tipos de datos y valores nulos

```python
estructura_df = pd.DataFrame({
    "tipo_dato": titanic_df.dtypes.astype(str),
    "valores_nulos": titanic_df.isna().sum(),
})
display(estructura_df)
```

## Generación automática del resumen

El resumen se construye desde el mismo DataFrame revisado arriba. La carpeta de resultados se crea si no existe y el archivo se reemplaza en cada ejecución.

```python
project_root = Path.cwd()
if not (project_root / "AE2").exists():
    project_root = project_root.parents[1]
output_dir = project_root / "AE2" / "resultados"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "resumen_dataframe.txt"

resumen = [
    "RESUMEN DEL DATAFRAME TITANIC",
    "============================",
    f"Identificador del dataset: {DATASET_ID}",
    f"Enlace: {DATASET_URL}",
    f"Fecha de consulta: {FECHA_CONSULTA}",
    f"Split utilizado: {SPLIT}",
    "Muestra utilizada: no; se cargó el split completo",
    f"Cantidad de filas: {titanic_df.shape[0]}",
    f"Cantidad de columnas: {titanic_df.shape[1]}",
    "",
    "COLUMNAS TIPOS Y VALORES NULOS",
    "--------------------------------",
    estructura_df.to_string(),
]
output_path.write_text("\n".join(resumen) + "\n", encoding="utf-8")

contenido_generado = output_path.read_text(encoding="utf-8")
print(f"Archivo generado: {output_path.relative_to(project_root)}")
print(contenido_generado)
```
